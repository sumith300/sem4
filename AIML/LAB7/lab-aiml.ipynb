```python
import numpy as np

def step(x):
    return 1 if x >= 0 else 0

def train_perceptron(X, y, learning_rate=0.1, epochs=10):
    weights = np.zeros(X.shape[1])
    bias = 0

    print("\nTraining Start")
    for epoch in range(epochs):
        print(f"\nEpoch {epoch+1}")
        for i in range(len(X)):
            z = np.dot(X[i], weights) + bias
            pred = step(z)
            error = y[i] - pred
            weights += learning_rate * error * X[i]
            bias += learning_rate * error
            print(f" Input: {X[i]}, Predicted: {pred}, Actual: {y[i]}, Error: {error}, Weights: {weights}, Bias: {bias}")
    print("Training Complete\n")
    return weights, bias

def predict(X, weights, bias):
    return [step(np.dot(x, weights) + bias) for x in X]
```

```python
print("AND Gate")
X_and = np.array([[0,0], [0,1], [1,0], [1,1]])
y_and = np.array([0, 0, 0, 1])
w_and, b_and = train_perceptron(X_and, y_and)
print("Predictions:", predict(X_and, w_and, b_and))

print("\nOR Gate")
X_or = np.array([[0,0], [0,1], [1,0], [1,1]])
y_or = np.array([0, 1, 1, 1])
w_or, b_or = train_perceptron(X_or, y_or)
print("Predictions:", predict(X_or, w_or, b_or))

print("\nNOT Gate")
X_not = np.array([[0], [1]])
y_not = np.array([1, 0])
w_not, b_not = train_perceptron(X_not, y_not)
print("Predictions:", predict(X_not, w_not, b_not))
```

## Perceptron Training Step Explanation for AND Gate

### Output Line:
Input: [0 0], Predicted: 1, Actual: 0, Error: -1, Weights: [0. 0.], Bias: -0.1

---

### What Each Part Means:

#### Input: [0 0]  
This is the first input in the AND gate truth table.  
Both inputs are 0, so the expected output is 0.

#### Predicted: 1  
The perceptron calculated the total input using the initial weights and bias:  
z = (0 × 0) + (0 × 0) + 0 = 0  
Then applied the step activation function:  
step(0) = 1 (since step(x) returns 1 when x >= 0)  
So, the perceptron predicted 1, which is incorrect for this input.

#### Actual: 0  
The correct output for [0, 0] in an AND gate is 0.

#### Error: -1  
The error is calculated as:  
error = actual - predicted = 0 - 1 = -1  
This negative error tells the model it was too confident and needs correction.

---

### Weight and Bias Update:

Learning rate: 0.1  
Input: [0, 0]  
Error: -1

Update rule:  
weights = weights + learning_rate × error × input  
         = [0, 0] + 0.1 × (-1) × [0, 0] = [0, 0]  
(Since the input is [0, 0], weights do not change)

Bias update:  
bias = bias + learning_rate × error  
     = 0 + 0.1 × (-1) = -0.1

---

### Final Values After This Step:
- Weights: [0. 0.] (unchanged)  
- Bias: -0.1 (decreased slightly)

---

## What Happened in This Training Step

The perceptron received no signal (both inputs were 0) but still predicted 1 due to a neutral bias.  
Since the prediction was incorrect, it learned to be slightly more conservative by reducing the bias.  
This makes it less likely to wrongly output 1 in similar low-input situations during future steps.

The weights remained unchanged because the inputs had no contribution ([0, 0] × anything = [0, 0]).  
Only the bias was adjusted.

This is how the perceptron learns: by minimizing error step by step, gradually updating its internal parameters to match the expected output more accurately over time.
