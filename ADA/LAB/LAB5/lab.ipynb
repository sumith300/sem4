```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import exposure
```

```python
eye_img = cv2.imread('eye.jpg')
low_img = cv2.imread('low.jpg')

eye_img = cv2.cvtColor(eye_img, cv2.COLOR_BGR2RGB)
low_img = cv2.cvtColor(low_img, cv2.COLOR_BGR2RGB)
```

```python
eye_gray = cv2.cvtColor(eye_img, cv2.COLOR_RGB2GRAY)
low_gray = cv2.cvtColor(low_img, cv2.COLOR_RGB2GRAY)
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(eye_img)
plt.title('Eye Image (Color)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(low_img)
plt.title('Low Contrast Image (Color)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(eye_gray, cmap='gray')
plt.title('Eye Image (Grayscale)')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(low_gray, cmap='gray')
plt.title('Low Contrast Image (Grayscale)')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 3, 1)
plt.imshow(eye_img)
plt.title('Eye Image (Color)')
plt.axis('off')

plt.subplot(2, 3, 2)
for i, color in enumerate(['red', 'green', 'blue']):
    plt.hist(eye_img[:,:,i].ravel(), bins=256, color=color, alpha=0.7)
plt.title('Eye Image - Color Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(2, 3, 3)
plt.hist(eye_gray.ravel(), bins=256, color='gray', alpha=0.7)
plt.title('Eye Image - Grayscale Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(2, 3, 4)
plt.imshow(low_img)
plt.title('Low Contrast Image (Color)')
plt.axis('off')

plt.subplot(2, 3, 5)
for i, color in enumerate(['red', 'green', 'blue']):
    plt.hist(low_img[:,:,i].ravel(), bins=256, color=color, alpha=0.7)
plt.title('Low Contrast - Color Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(2, 3, 6)
plt.hist(low_gray.ravel(), bins=256, color='gray', alpha=0.7)
plt.title('Low Contrast - Grayscale Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

## Original Images Analysis
The histograms show different contrast levels between the images. The eye image has better distribution across pixel intensities, while the low contrast image has pixels concentrated in specific intensity ranges.

```python
eye_gray_eq = cv2.equalizeHist(eye_gray)
low_gray_eq = cv2.equalizeHist(low_gray)
```

```python
eye_hsv = cv2.cvtColor(eye_img, cv2.COLOR_RGB2HSV)
eye_hsv[:,:,2] = cv2.equalizeHist(eye_hsv[:,:,2])
eye_color_eq = cv2.cvtColor(eye_hsv, cv2.COLOR_HSV2RGB)

low_hsv = cv2.cvtColor(low_img, cv2.COLOR_RGB2HSV)
low_hsv[:,:,2] = cv2.equalizeHist(low_hsv[:,:,2])
low_color_eq = cv2.cvtColor(low_hsv, cv2.COLOR_HSV2RGB)
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 4, 1)
plt.imshow(eye_gray, cmap='gray')
plt.title('Eye - Original Grayscale')
plt.axis('off')

plt.subplot(2, 4, 2)
plt.hist(eye_gray.ravel(), bins=256, color='gray', alpha=0.7)
plt.title('Original Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(2, 4, 3)
plt.imshow(eye_gray_eq, cmap='gray')
plt.title('Eye - Equalized Grayscale')
plt.axis('off')

plt.subplot(2, 4, 4)
plt.hist(eye_gray_eq.ravel(), bins=256, color='gray', alpha=0.7)
plt.title('Equalized Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(2, 4, 5)
plt.imshow(low_gray, cmap='gray')
plt.title('Low - Original Grayscale')
plt.axis('off')

plt.subplot(2, 4, 6)
plt.hist(low_gray.ravel(), bins=256, color='gray', alpha=0.7)
plt.title('Original Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(2, 4, 7)
plt.imshow(low_gray_eq, cmap='gray')
plt.title('Low - Equalized Grayscale')
plt.axis('off')

plt.subplot(2, 4, 8)
plt.hist(low_gray_eq.ravel(), bins=256, color='gray', alpha=0.7)
plt.title('Equalized Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 4, 1)
plt.imshow(eye_img)
plt.title('Eye - Original Color')
plt.axis('off')

plt.subplot(2, 4, 2)
for i, color in enumerate(['red', 'green', 'blue']):
    plt.hist(eye_img[:,:,i].ravel(), bins=256, color=color, alpha=0.7)
plt.title('Original Color Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(2, 4, 3)
plt.imshow(eye_color_eq)
plt.title('Eye - Equalized Color')
plt.axis('off')

plt.subplot(2, 4, 4)
for i, color in enumerate(['red', 'green', 'blue']):
    plt.hist(eye_color_eq[:,:,i].ravel(), bins=256, color=color, alpha=0.7)
plt.title('Equalized Color Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(2, 4, 5)
plt.imshow(low_img)
plt.title('Low - Original Color')
plt.axis('off')

plt.subplot(2, 4, 6)
for i, color in enumerate(['red', 'green', 'blue']):
    plt.hist(low_img[:,:,i].ravel(), bins=256, color=color, alpha=0.7)
plt.title('Original Color Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.subplot(2, 4, 7)
plt.imshow(low_color_eq)
plt.title('Low - Equalized Color')
plt.axis('off')

plt.subplot(2, 4, 8)
for i, color in enumerate(['red', 'green', 'blue']):
    plt.hist(low_color_eq[:,:,i].ravel(), bins=256, color=color, alpha=0.7)
plt.title('Equalized Color Histogram')
plt.xlabel('Pixel Intensity')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

## Histogram Equalization Results
Histogram equalization redistributes pixel intensities to improve contrast. The equalized images show better detail visibility and more uniform histogram distribution across all intensity levels.

```python
plt.figure(figsize=(20, 5))

plt.subplot(1, 4, 1)
plt.imshow(eye_gray, cmap='gray')
plt.title('Eye - Original Grayscale')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(eye_gray_eq, cmap='gray')
plt.title('Eye - Equalized Grayscale')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(low_gray, cmap='gray')
plt.title('Low Contrast - Original Grayscale')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(low_gray_eq, cmap='gray')
plt.title('Low Contrast - Equalized Grayscale')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(20, 5))

plt.subplot(1, 4, 1)
plt.imshow(eye_img)
plt.title('Eye - Original Color')
plt.axis('off')

plt.subplot(1, 4, 2)
plt.imshow(eye_color_eq)
plt.title('Eye - Equalized Color')
plt.axis('off')

plt.subplot(1, 4, 3)
plt.imshow(low_img)
plt.title('Low Contrast - Original Color')
plt.axis('off')

plt.subplot(1, 4, 4)
plt.imshow(low_color_eq)
plt.title('Low Contrast - Equalized Color')
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Conclusion
Histogram equalization effectively enhances image contrast by redistributing pixel intensities. The low contrast image shows significant improvement in detail visibility after equalization. Both grayscale and color images benefit from this technique, with color images processed through the HSV color space to preserve color information while enhancing luminance.
