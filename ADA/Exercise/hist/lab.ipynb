```python
from skimage import io, color
from skimage.measure import shannon_entropy
import cv2
```

```python
import numpy as np
from matplotlib import pyplot as plt

image_rgb=cv2.imread('hist.jpg')
histogram=cv2.calcHist([image_rgb],[0],None,[256],[0,256])
```

```python
plt.plot(histogram)
plt.title(' Gray scale Histogram')
plt.xlabel('Pixel Value') #pixel value
plt.ylabel('Frequency') #pixel count
plt.xlim([0, 256])
plt.show()
```

```python
newImage=cv2.imread('cham.jpg')
newImage=cv2.cvtColor(newImage, cv2.COLOR_BGR2GRAY)
# plt.imshow(newImage)
```

```python
colors=('b','g','r')
plt.figure()
for i, color in enumerate(colors):
    histogram=cv2.calcHist([newImage],[i],None,[256],[0,256])
    plt.plot(histogram, color=color)
    plt.xlim([0, 256])
plt.show()    
```

```python


def display_images_and_histograms(path):
    original=cv2.imread(path, cv2.IMREAD_GRAYSCALE)

    equalized = cv2.equalizeHist(original)
    fig, axes = plt.subplots(2, 2, figsize=(12, 6))
    
    axes[0, 0].imshow(original, cmap='gray')
    axes[0, 0].set_title('Original Image')
    axes[0, 0].axis('off')
    
    axes[0, 1].imshow(equalized, cmap='gray')
    axes[0, 1].set_title('Equalized Image')
    axes[0, 1].axis('off')

    axes[1, 0].hist(original.ravel(), bins=256, range=[0,256])
    axes[1, 0].set_title('Histogram of Original Image')

    axes[1, 1].hist(equalized.ravel(), bins=256, range=[0,256])
    axes[1, 1].set_title('Histogram of Equalized Image')

    plt.tight_layout()
    plt.show()
    

display_image_and_histogram('hist.jpg')
```
